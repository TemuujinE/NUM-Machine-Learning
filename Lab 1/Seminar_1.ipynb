```python
import re
import random
import string
import textwrap

import numpy as np
import pandas as pd

import warnings
warnings.filterwarnings('ignore')
```

# 1. Өгөгдсөн хүснэгтээс нийт ялгаатай утгатай элемэнтүүдийн тоог ол.

```python
def generate_random_string(length):
    return ''.join(random.choices(string.ascii_letters + string.digits, k = length))

def generate_random_dataframe(n):
    df = pd.DataFrame({'Col1': [generate_random_string(10) for _ in range(n)],
                       'Col2': [generate_random_string(10) for _ in range(n)]})
    return df
```

```python
df = generate_random_dataframe(n = 30)

print("First col:")
print(df['Col1'].value_counts().sample(10))

print("\nSecond col:")
print(df['Col2'].value_counts().sample(10))
```

# 2. Өгөгдсөн өгүүлбэрийн үг бүр нь хэдэн удаа давтагдсанг ол.

```python
lorem_ipsum = """Lorem Ipsum is simply dummy text of the printing and 
                 typesetting industry Lorem Ipsum has been the industry's standard dummy 
                 text ever since the 1500s, when an unknown printer took a galley of type and 
                 scrambled it to make a type specimen book It has survived not only five centuries, 
                 but also the leap into electronic typesetting, remaining essentially unchanged 
                 It was popularised in the 1960s with the release of Letraset sheets containing 
                 Lorem Ipsum passages, and more recently with desktop publishing software 
                 like Aldus PageMaker including versions of Lorem Ipsum."""
```

```python
# 1. No puncts
text_no_punct = re.sub(r'[^\w\s\']', '', lorem_ipsum)
print(textwrap.fill(text_no_punct))
```

```python
# 2. 's and s after numbers
text_no_suffix = re.sub(r'(?<=\d)s\b', '', text_no_punct)
print(textwrap.fill(text_no_suffix))
```

```python
# 3. Remove ' letters
text_no_apos = re.sub(r'\b\w+\'\w+\b', '', text_no_suffix)
print(textwrap.fill(text_no_apos))
```

```python
# 4. contractions like e.g 're, 've
text_no_contractions = re.sub(r'\b\w+\'(?:re|ve|s|m|d|ll)\b', '', text_no_apos)
print(textwrap.fill(text_no_contractions))
```

```python
# 5. No extra space
clean_text = re.sub(r'\s+', ' ', text_no_contractions).strip()
print(textwrap.fill(clean_text))
```

```python
word_cnt = pd.DataFrame({"words": [word.strip() for word in clean_text.split()]})

print("Each word cnt in sentence:")
print(word_cnt['words'].value_counts())
```

# 3. Өгөгдсөн хүснэгтээс 'xyz' тэмдэгт мөрийг агуулсан тэмдэгт мөрүүдийг хэвлэ.

```python
def generate_random_string(length):
    prefix_length = random.randint(0, length - 3)
    suffix_length = length - prefix_length - 3
    
    xyz_pattern = 'xyz'
    prefix = ''.join(random.choices(string.ascii_letters + string.digits, k = prefix_length))
    suffix = ''.join(random.choices(string.ascii_letters + string.digits, k = suffix_length))

    random_number = np.random.randint(6)
    if random_number % 3 == 0:
        return prefix + xyz_pattern + suffix
    return prefix + suffix
```

```python
xyz_objects = pd.DataFrame({'Col': [generate_random_string(15) for _ in range(100)]})
xyz_objects.sample(15)
```

```python
print("Column values containing xyz:")
print(xyz_objects[xyz_objects['Col'].str.contains('xyz')].reset_index(drop = True))
```

# 4. Өгөгдсөн хүснэгтээс 'xyz' тэмдэгт мөрөөр эхэлсэн эсвэл төгссөн тэмдэгт мөрүүдийг хэвлэ.

```python
xyz_objects = pd.DataFrame({'Col': [generate_random_string(15) for _ in range(1000)]})
xyz_objects.sample(15)
```

```python
print("Column values startswith or endswith:")
print(xyz_objects[(xyz_objects['Col'].str.endswith('xyz')) | 
                  (xyz_objects['Col'].str.startswith('xyz'))].reset_index(drop = True))
```

# 5. Хамгийн эхний анхны 50 тоог олох функц бич.

```python
def sieve(n):
    primes = []
    is_prime = [True] * (n * 10)

    for num in range(2, int(n * (n**0.5)) + 1):
        if is_prime[num]:
            primes.append(num)
            for multiple in range(num * num, n * 10, num):
                is_prime[multiple] = False

    for num in range(max(2, int(n * (n**0.5)) + 1), n * 10):
        if is_prime[num]:
            primes.append(num)

    return primes[:n]
```

```python
print("First 50 primes:")
print(textwrap.fill(', '.join(str(num) for num in sieve(50))))
```
